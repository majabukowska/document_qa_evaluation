```python
from bs4 import BeautifulSoup
import json
```

```python
path = "/Users/majabukowska/PycharmProjects/document_qa_evaluation/data/raw/What is PyTorch_ _ IBM.html"
with open(path, "r", encoding="utf-8") as f:
    html_text = f.read()
```

```python
soup = BeautifulSoup(html_text, "html.parser")

for tag in soup(["nav", "footer", "aside", "script", "style"]):
    tag.decompose()

text = soup.get_text(separator="\n")
```

```python
lines = text.splitlines()

clean_lines = [
    line.strip()
    for line in lines
    if line.strip() != ""
]

clean_text = "\n".join(clean_lines)
```

```python
blacklist_phrases = [
    "This website utilizes technologies such as cookies",
    "Privacy Policy",
    "Manage Preferences",
    "Learn",
    "Get Started",
    "Tutorials",
    "opens in a new window",
    "close this dialog",
    "cookie preferences",
    "privacy policy",
    "edit on github",
    "show source",
    "view docs",
    "community",
    "newsletter",
]

for phrase in blacklist_phrases:
    clean_text = clean_text.replace(phrase, "")
```

```python
def chunk_text(text, max_size=1000):
    chunks = []
    start = 0
    while start < len(text):
        end = start + max_size
        chunk = text[start:end]
        chunks.append(chunk)
        start = end
        
    return chunks
    
    
```

```python
chunks = chunk_text(clean_text, max_size=1000)
len(chunks)
```

```python
processed_path = "../data/processed/chunks.json"

data = [
    {
    "id": i,
    "text": chunk,
    "source": "torch.nn documentation"
    } for i, chunk in enumerate(chunks)
]

with open(processed_path, "w", encoding="utf-8") as f:
    json.dump(data, f, ensure_ascii=False, indent=2)
```
